```python
df['price category']= pd.cut(
    df['price'],
    bins=[-1,100,200,float('inf')],
    labels=['Budget','Standard','Premium']
)
df[['price','price category']].head(10)
```

```python
df['avaliablility_category']=pd.cut(
    df['availability_365'],
    bins=[-1,50,180,365],
    labels=['Low','Medium','High']
)
df[['avaliablility_category','availability_365']].head(20)
```

```python
df['host_type']=np.where(
    df['calculated_host_listings_count']==1,
    'Single-listing Host',
    'Multi-listing Host'
)
df[['host_type','calculated_host_listings_count']].head(10)
```

```python
df['review category']=pd.cut(
    df['number_of_reviews'],
    bins=[-1,0,10,50,float('inf')],
    labels=['No Review','Few Review','Moderate Review','Many Review']
)
df[['review category','number_of_reviews']].head(10)
```

```python
plt.figure(figsize=(5, 3))

plt.hist(df['price'].dropna(), bins=50)
plt.xlim(0,800)

plt.title('Distribution of Airbnb Prices')
plt.xlabel('Price per Night ($)')
plt.ylabel('Number of Listings')

#plt.tight_layout()
plt.show()
```

```python
df.columns.tolist()
```

```python
#listings fall into each price category
price_count=df['price category'].value_counts().reindex(['Budget','Standard','Premium'])
plt.figure(figsize=(5,3))
plt.bar(price_count.index.astype(str),price_count.values,color='purple')
#plt.ylim(0,2000)
plt.title('Airbnb by Price Category')
plt.xlabel('Price Category')
plt.ylabel('Number of listing')
plt.tight_layout()
plt.show()
```

```python
#listings fall into each room category
room_category=df['room_type'].value_counts()
plt.figure(figsize=(5,3))
plt.bar(room_category.index,room_category.values,color='pink')
plt.xlabel('Room Category')
plt.ylabel('Number of listing')
plt.tight_layout()
plt.show()
```

```python
#Which room type has the highest average nightly price
avg_price =df.groupby('room_type')['price'].mean().sort_values(ascending=False)
plt.figure(figsize=(5,3))
plt.bar(avg_price.index,avg_price.values)
plt.xlabel('Room type')
plt.ylabel('Avg price per room')
plt.tight_layout()
plt.show()
```

```python
avaliability_count=df['avaliablility_category'].value_counts().reindex([
    'High','Medium','Low'
],fill_value=0)
plt.figure(figsize=(5,5))
plt.pie(
    avaliability_count.values,
    labels=avaliability_count.index,
    autopct='%1.1f%%',
    startangle=90)
plt.title('Avaliable Category')
plt.tight_layout()
plt.show()
```
